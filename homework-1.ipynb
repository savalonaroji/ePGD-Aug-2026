```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5 * np.pi, 5 * np.pi, 100)
y = x
plt.plot(x, y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5 * np.pi, 5 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5 * np.pi, 5 * np.pi, 100)
y = x * np.sin(x)
plt.plot(x, y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5 * np.pi, 5 * np.pi, 500)

plt.figure(figsize=(6.5, 4.8))
plt.plot(x, x, color='blue', label='x', linewidth=1)
plt.plot(x, -x, color='blue', label='-x', linewidth=1)
plt.plot(x, np.sin(x), color='green', label='sin(x)', linewidth=1.5)
plt.plot(x, x * np.sin(x), color='red', label='xsin(x)', linewidth=2.5)

plt.annotate('origin', xy=(0, 0), xytext=(0, 0))

plt.xlim(-5 * np.pi, 5 * np.pi)
plt.ylim(-5 * np.pi, 5 * np.pi)

plt.tick_params(top=True, right=True, direction='in')
plt.legend(loc='upper right', framealpha=1, edgecolor='black', fancybox=False, fontsize=11.5)
plt.show()
```
